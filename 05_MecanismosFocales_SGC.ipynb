```python
cvs_mecanismos_focales = 'Cat_mec_foc_CC_2014-2024_v2.csv'

df = pd.read_csv(os.path.join(path_datasets, cvs_mecanismos_focales),
                 sep=';',
                 decimal=',')

df.head()
```

```python
df_mf = df[['FECHA_HORA_UTC',
            'Numero',
            'LATITUD (°) LOCALIZACIÓN HIPOCENTRO',
            'LONGITUD (°) LOCALIZACIÓN HIPOCENTRO',
            'PROFUNDIDAD (Km) LOCALIZACIÓN HIPOCENTRO',
            'MAGNITUD',
            'Strike plano 1 (°) PLANOS NODALES',
            'Dip plano 1 (°) PLANOS NODALES',
            'Rake plano 1 (°) PLANOS NODALES',
            'Strike plano 2 (°) PLANOS NODALES',
            'Dip plano 2 (°) PLANOS NODALES',
            'Rake plano 2 (°) PLANOS NODALES',
            'Catalogo']].copy()
df_mf. columns = ['Fecha',
                  'Numero',
                  'Latitud',
                  'Longituf',
                  'Profundidad',
                  'Ml',
                  'Strike plano 1',
                  'Dip plano 1',
                  'Rake plano 1',
                  'Strike plano 2',
                  'Dip plano 2',
                  'Rake plano 2',
                  'Metodo']

df_mf.head()
```

```python
color_line = {'SWIFT': 'tab:blue',
              'SCMTV': 'tab:brown',
              'ISOLA': 'tab:green',
              'FASE W': 'tab:red',
              'POLARIDADES': 'tab:purple'}
```

```python
df_mf['Profundidad'].unique()
```

## Plot

```python
for n, strike, dip, rake, met, prof in zip(df_mf['Numero'],#[0:1],
                                     df_mf['Strike plano 1'],
                                     df_mf['Dip plano 1'],
                                     df_mf['Rake plano 1'],
                                     df_mf['Metodo'],
                                     df_mf['Profundidad']):
  focalmechs = [strike, dip, rake] # strike, dip, and rake
  nombre = f'{n}_{met}'
  print(nombre)

  if prof <= 10:
    color_ = 'b'
  elif prof <= 20:
    color_ = 'g'
  elif prof <= 40:
    color_ = 'y'
  elif prof <= 60:
    color_ = 'r'

  beachball(focalmechs,
            facecolor=color_,
            edgecolor=color_line[met],
            linewidth=7,
            outfile=os.path.join(path_save_figures,f'{nombre}.svg'),
            format='svg')



  print('\n')
```
