# Cargar catalogos

```python
cvs_mecanismos_focales = 'MF_5km_Cluster_12.csv'

df = pd.read_csv(os.path.join(path_datasets, cvs_mecanismos_focales),
                 sep=';',
                 decimal=',')

df['FechaHora'] = pd.to_datetime(
    df['FechaHora'].astype(int).astype(str),
    format='%Y%m%d%H%M'
)

df.sort_values(by='FechaHora', inplace=True, ascending=True)
df.reset_index(drop=True, inplace=True)

df.info()
```

```python
df.head()
```

```python
df.columns
```

```python
fig, ax = plt.subplots(figsize=(5.5, 10), nrows=4, ncols=3)
ax = ax.flatten()
n_ax = 0
for fecha, strike, dip, rake, prof, nst, gap, ml in zip(df['FechaHora'],
                                          df['STRIK'],
                                          df['DIP'],
                                          df['RAKE'],
                                          df['Depth'],
                                          df['NST'],
                                          df['GAP'],
                                          df['Ml'],):
  focalmechs = [strike, dip, rake] # strike, dip, and rake


  b = beach(focalmechs,
            linewidth=3,
            width=400,
            axes=ax[n_ax])
  ax[n_ax].add_collection(b)
  ax[n_ax].set_xlim(-1, 1)
  ax[n_ax].set_ylim(-1, 1)
  ax[n_ax].set_aspect('equal')

  ax[n_ax].set_title(f'{fecha}\nProf. = {prof:.1f} km\n Ml = {ml:.1f} \nNST = {nst}\nGAP = {gap}°',
                      fontsize=9)
  ax[n_ax].set_xticks([])
  ax[n_ax].set_yticks([])
  n_ax += 1

plt.tight_layout()

plt.savefig(os.path.join(path_save_figures,'MF_cluster_12.png'),
            dpi=300,
            bbox_inches='tight')

plt.show()
```
